```python
#Brazil has won the highest number of FIFA World Cups.

wins=df["Winner"].value_counts()
wins
```

```python
wins.plot(kind="bar")
plt.title("FIFA WORLD CUP WIN BY COUNTRY")
plt.xlabel("Country")
plt.ylabel("Number of World Cup")
plt.show()
```

```python
df["Runners-Up"].value_counts()
```

```python
runner=df["Runners-Up"].value_counts()
runner.plot(kind="bar")
plt.title("Runner-Up Appearences")
plt.xlabel("Country")
plt.ylabel("NUmber of Times")
plt.show()
```

```python
# Tournament with the Highest Goals

df[df["GoalsScored"]==df["GoalsScored"].max()]
```

```python
# Highest Attendance

df[df["Attendance"]==df["Attendance"].max()]
```

```python
# The Average Goals

df["GoalsScored"].mean()
```

```python
#Goals Scored Over the Years
plt.figure(figsize=(10,5))
plt.plot(df["Year"],df["GoalsScored"],marker="o")
plt.title("Goals Scored in FIFA World Cups")
plt.xlabel("Year")
plt.ylabel("Goals Scored")
plt.grid()
plt.show()
```

```python
#Attendance Over the Years
plt.figure(figsize=(10,5))
plt.plot(df["Year"],df["Attendance"],marker='o')

plt.title("World Cup Attendance")
plt.xlabel("Year")
plt.ylabel("Attendance")
plt.grid()
plt.show()
```

```python
#Qualified Teams

plt.figure(figsize=(10,5))

plt.bar(df["Year"],df["QualifiedTeams"])
plt.title("Qualified Teams")
plt.xlabel("Year")
plt.ylabel("Number of teams")
plt.show()
```

```python
#Match Played
plt.figure(figsize=(10,5))

plt.plot(df["Year"], df["MatchesPlayed"], marker="o")

plt.title("Matches Played")
plt.xlabel("Year")
plt.ylabel("Matches")

plt.grid()

plt.show()
```

```python
#world cup winners

wins=df["Winner"].value_counts()

plt.figure(figsize=(8,8))
plt.pie(
    wins,
    labels=wins.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of World Cup Titles")

plt.show()
```

```python
#Histogram of Goals
plt.figure(figsize=(8,5))

plt.hist(df["GoalsScored"], bins=8)

plt.title("Distribution of Goals")
plt.xlabel("Goals")
plt.ylabel("Frequency")

plt.show()
```
